## Project Overview

This semester, my partner Najihah Ahmad Fikri and I built a full-stack interactive dashboard called **NYC Building Insights** as part of Advanced Computing for Policy coursed at SIPA, Columbia. The dashboard explores building activity data across New York City's five boroughs, combining six public datasets from NYC Open Data into a unified, automated pipeline.

The project gave us hands-on experience with the full data engineering lifecycle: extracting data from public APIs, loading it into a cloud data warehouse, automating daily refresh with CI/CD, and building an interactive web dashboard on top of it.

## How the Project Started and Evolved

Our original proposal focused on a straightforward question: is there a relationship between eviction rates and median household income across NYC boroughs? We planned to use two datasets: NYC Evictions and ACS Census income data.

Through feedback and iteration, we realized the scope was too narrow to build a meaningful dashboard. We expanded to include building permits, complaints, facade inspections, and building footprints. The project evolved from a simple eviction analysis into a comprehensive view of how NYC buildings are built, maintained, and sometimes lost.

By the end, the dashboard covered six datasets, four pages, and an automated daily refresh pipeline, far beyond what we originally scoped.

## What We Built

The dashboard has four main pages:

**Buildings Overview** explores NYC's total building stock by borough, building density per square mile, new construction activity from 2008 to 2020, active construction jobs from 2025 onwards, and facade safety inspection data. It includes an interactive year slider, gauge charts for facade safety rates, and a cycle slider that filters all facade visualizations simultaneously.

**Building Evictions** tracks eviction trends from 2017 onwards, with residential versus commercial breakdowns, seasonal patterns, a year range slider, and an anomaly detection chart that flags months with unusually high eviction activity relative to the historical mean.

**Building Complaints** analyzes complaints filed with the NYC Department of Buildings from 2015 onwards. It includes priority classification based on DOB codes, response time analysis, and trend charts by borough and category.

**Proposal** documents the project's evolution from initial research questions to the final dashboard.

## The Data Pipeline

One of the most technically interesting parts of the project was building the ETL pipeline. We pull data from six NYC Open Data APIs, process and standardize it in Python, and load it into Google BigQuery using a truncate-and-replace method. The pipeline runs automatically every day at 6am UTC via GitHub Actions.

We chose the truncate method because BigQuery's free tier does not support DML operations like DELETE or UPDATE. Every run drops and recreates each table with fresh data, which keeps things simple and avoids duplicates.

The Streamlit app reads from BigQuery using `pandas_gbq` with `st.cache_data(ttl=3600)` caching. The first page load hits BigQuery and takes a few seconds. Every subsequent load within the hour is near-instant because Streamlit serves the cached result.

We ran into several unexpected challenges during the pipeline build:

- The building footprints dataset stores geometry as polygons, which the JSON API cannot return. We had to switch to the CSV endpoint to bypass this limitation.
- The DOB permit datasets span two separate systems with different column names, date formats, and coverage periods that required careful normalization before combining.
- NYC Open Data API responses sometimes have null dates or inconsistent string formats that break standard `pd.to_datetime()` parsing, requiring `format='mixed'` and error coercion.

## What I Learned

This project taught me that real-world data is messier than it looks in documentation. Column names differ from what the API documentation says. Date formats are inconsistent across datasets from the same city agency. Fields that appear to be numeric are sometimes stored as strings. Each of these required debugging and iteration.

I also learned how to think about performance at scale. Filtering in SQL before loading into Python, caching query results, and aggregating at the source rather than in the app made a significant difference in how fast the dashboard loads.

Working with BigQuery, GitHub Actions, and Streamlit Cloud together gave me hands-on experience with a realistic data engineering stack. The combination of automated refresh, cloud storage, and a live web interface is something I can apply directly to policy analysis work.

Finally, building for a real audience changed how I thought about design. The goal was to make findings accessible and interpretable to someone without a data background, which pushed me to add context panels, plain-language labels, and visual hierarchy to every page.
