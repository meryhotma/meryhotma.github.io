# Step 1 : Importing The Data

```python
# import necessary packages

import pandas as pd
import plotly.express as px

# import NYC park data

park = pd.read_csv("Parks_Properties_20251205.csv", delimiter=",")

park
```

```python
# import NYC Park toilet data

toilet = pd.read_csv("Directory_Of_Toilets_In_Public_Parks_20251205.csv", delimiter=",")

toilet
```

# Step 2a - Data Cleaning and Formatting - Park Dataset
we need to do some data handling to ensure that each data set has the same format

Columns that are likely to be used in park dataset, and how we handle them:
- Borough : Change the code in "Borough" series according to [this data dictionary]("https://docs.google.com/spreadsheets/d/1Q4DBWu7riNFxWvy1vnTJHoOI3r2L9oW6eCN56jCNyCw/edit?gid=0#gid=0"")
- TYPECATEGORY = The type of property according to Planning and Development Division of the New York City Department of Parks and Recreation. [see]("https://docs.google.com/document/d/1NExNJF5YKID04oOopi0fHainRuGG3Pz_jKSrMujPsPk/edit?tab=t.0")
- Acres = Size of the park property (in acres).

```python
# check null value in borough
park["BOROUGH"].info()
```

```python
# check unique data in Borough
park["BOROUGH"].unique()
```

```python
BORO = []

for code in park["BOROUGH"]:
    if code == "B":
        BORO.append("Brooklyn")
    elif code == "M":
        BORO.append("Manhattan")
    elif code == "Q":
        BORO.append("Queens")
    elif code == "R":
        BORO.append("Staten Island")
    elif code == "X":
        BORO.append("Bronx")
    else:
        BORO.append("Unknown")

# add the list to data frame
park["BORO"] = BORO
# check the result
print(park["BOROUGH"].unique())
print(park["BORO"].unique())
park
```

```python
# Handling Area Data

park["ACRES"].info()
```

The acres column is in string, so we need to convert it to float

```python
# change acre data type to folat, and omit comas
park["ACRES"] = park["ACRES"].str.replace(",", "").astype(float)

park["ACRES"].info()
```

let's check the Rercreational Spot type

```python
# check TYPECATEGORY series info
park["TYPECATEGORY"].info()

# check TYPECATEGORY unique values
park["TYPECATEGORY"].unique()
```

we need to filter the park data to only take park category that is possible for recreation, mainly (but not all) managed by NYCPARKS. (I'm open to every suggestion to add or omit some categories)

```python
# set the valid Categories
valid_categories = [
    "Community Park",
    "Neighborhood Park",
    "Playground",
    "Flagship Park",
    "Jointly Operated Playground",
    "Recreational Field/Courts",
    "Triangle/Plaza",
    "Waterfront Facility",
    "Nature Area",
]

# filter TYPE CATEGORY by valid category
park_filtered = park[park["TYPECATEGORY"].isin(valid_categories)]

park_filtered
```

all datas in park dataset are classified as park

# Step 3a: Analysis on Number of Parks Per Borough

1. Number of Parks Per Borough

```python
# calculate number of park per borough
park_count = park_filtered.groupby("BORO").size().reset_index(name="TOTAL_PARK")

park_count
```

Finding 1 = Brooklyn has the highest number of recreational parks, while Staten Island has the fewest. This indicates that Brooklyn offers the greatest number of park options for residents, whereas Staten Island has comparatively limited park availability.

2. Average Area of Park of Each Borough

```python
# calculate the average area of of park per borough
park_area = (
    park_filtered.groupby("BORO")["ACRES"].mean().reset_index(name="AVG_AREA_OF_PARK")
)

park_area
```

Finding 2: Although Brooklyn contains the largest number of parks, these parks tend to be much smaller in size, resulting in the lowest average park area among the five boroughs. This suggests that Brooklyn’s parks may experience higher crowding and user density, especially during weekends and holiday seasons. The borough may require additional investment in park expansion or the creation of new green spaces to meet public demand.

3. Total Area of Park in Each Borough

```python
# calculate the average area of of park per borough
park_area_total = (
    park_filtered.groupby("BORO")["ACRES"].sum().reset_index(name="TOT_AREA_OF_PARK")
)

park_area_total.sort_values("TOT_AREA_OF_PARK")
```

Finding 3: Manhattan has the lowest area of park while the Staten Island has the largest area of park in NYC. This shows Manhattan may need more outdoor spaces in the future

3. Number of Parks VS Area of Parks-VIZ

We will compare the number of parks and area of parks with 2 visualization:
1. bar to represent number of park and scatter (point) to represent avg area of park
2. scatterplot to provide the corelation between number of park and total area of park 

```python
import plotly.io as pio

pio.renderers.default = "notebook_connected+plotly_mimetype"
```

```python
# merge both tables
park_number_area = pd.merge(park_count, park_area, on="BORO", how="left")

park_number_area

park_total = pd.merge(park_number_area, park_area_total, on="BORO", how="left")

park_total
```

```python
import plotly.graph_objects as go

fig = go.Figure()

# bar for total parks
fig.add_trace(
    go.Bar(x=park_total["BORO"], y=park_total["TOTAL_PARK"], name="Total Parks")
)

# points
fig.add_trace(
    go.Scatter(
        x=park_total["BORO"],
        y=park_total["TOT_AREA_OF_PARK"],
        name="Total Park Area (acres)",
        mode="markers",
        marker=dict(size=12, color="red"),
        yaxis="y2",
    )
)

fig.update_layout(
    title="Total Parks vs Total Park Area by Borough",
    xaxis_title="Borough",
    yaxis_title="Total Parks",
    yaxis2=dict(title="Total Park Area (acres)", overlaying="y", side="right"),
    template="plotly_white",
)

fig.show()
```

```python
# scatter plot to show correlation betewen number of park and areas of park

fig_scatter = px.scatter(
    park_total,
    x="TOTAL_PARK",
    y="TOT_AREA_OF_PARK",
    color="BORO",
    text="BORO",
    title="Total Number of Parks vs Total Park Area by Borough",
    labels={
        "TOTAL_PARK": "Total Parks",
        "TOT_AREA_OF_PARK": "Total Park Area (acres)",
        "BORO": "Borough",
    },
    trendline="ols",
)

# Make points bigger
fig_scatter.update_traces(marker=dict(size=18), textposition="top center")

fig_scatter.show()
```

Finding 4 the OLS trendline is not showing in this scatterplot, there is a high chance that there is low correlation between Total Park Areas and Number of Park of each Borough in NYC

# Step 3b. Analysis on Number of Toilet in Public Park of Each Borough

1. Number of Toilet in Public Park of Each Borough

```python
# calculate number of toilet in public park per borough
toilet_count = toilet.groupby("Borough").size().reset_index(name="total_toilet_park")
toilet_count.sort_values("total_toilet_park")
```

2. Visualization

```python
# number of toilet in graph
fig_t = px.bar(
    toilet_count,
    x="Borough",
    y="total_toilet_park",
    color="Borough",
    title="Number of Toilet in Public Park by Borough",
    labels={"Borough": "Borough", "total_toilet_park": "Total Toilet in Public Park"},
)
# sort the values
fig_t.update_layout(
    xaxis={
        "categoryorder": "array",
        "categoryarray": toilet_count.sort_values("total_toilet_park", ascending=False)[
            "Borough"
        ],
    }
)
fig_t.show()
```

Finding 5 = Brooklyn has the highest number of toilet in public park, and Staten Island has the lowest number.

# Step 4 Correlation Between Park and Number of Toilet in Public Park
In this step, we combine park dataset and the public toilet dataset to compare:
- the number of parks in each borough,
- the number of public toilets in each borough, and
- the number of toilets per acre of parkland.

This helps us understand whether access to public toilets is proportional to the amount of park space available in each borough.

```python
# combine both dataset
park_toilet = pd.merge(
    left=park_total, right=toilet_count, left_on="BORO", right_on="Borough", how="left"
)

park_toilet
```

Calculate the number of toilet per acres in NYC Park

```python
# Calculate the number of toilet per acres in NYC Park
park_toilet["toilet_per_park"] = (
    park_toilet["total_toilet_park"] / park_toilet["TOTAL_PARK"]
)

park_toilet.sort_values("toilet_per_park")
```

```python
# number of toilet per park
fig_1 = px.bar(
    park_toilet,
    x="Borough",
    y="toilet_per_park",
    color="Borough",
    title="Number of Toilets Per Park by Borough",
    labels={"Borough": "Borough", "toilet_per_park": "Total Toilet in Public Park"},
)
# sort the values
fig_1.update_layout(
    xaxis={
        "categoryorder": "array",
        "categoryarray": park_toilet.sort_values("toilet_per_park", ascending=False)[
            "Borough"
        ],
    }
)
fig_1.show()
```

Finding 6: Staten Island has the smallest number of Toilets per Park (apprx. 0.2 toilets per park) and Manhattan has the largest number of toilet per part (less than 0.5). This shows that there is less than 1 toilet on each park in all Borough in NYC, indicating that visitors may have to consider alternatives of restrooms in surrounding areas of the park.

Calculate number of toilet per acre

```python
# number of toilet per acre
park_toilet["toilet_per_acre"] = (
    park_toilet["total_toilet_park"] / park_toilet["TOT_AREA_OF_PARK"]
)

park_toilet.sort_values("toilet_per_acre")
```

```python
# number of public toilet in park per acre by borough
fig_2 = px.bar(
    park_toilet,
    x="Borough",
    y="toilet_per_acre",
    color="Borough",
    title="Number of Public Toilets in Park per Acre by Borough",
    labels={
        "Borough": "Borough",
        "toilet_per_acre": "Total Toilet in Public Park per Acre",
    },
)
# sort the values
fig_2.update_layout(
    xaxis={
        "categoryorder": "array",
        "categoryarray": park_toilet.sort_values("toilet_per_acre", ascending=False)[
            "Borough"
        ],
    }
)
fig_2.show()
```

finding 7: Brooklyn and Manhattan are two largest number of toilet per acre in NYC, and Staten Island is the lowest. Compared to the total area, the coverage of toilet in parks in NYC is less than 1 per acre, indicating that visitors may have to take a long walk before finding the nearest toilet in public park

Visualizing

```python
import plotly.graph_objects as go

fig = go.Figure()

# Bar for toilets per park
fig.add_trace(
    go.Bar(
        x=park_toilet["BORO"],
        y=park_toilet["toilet_per_park"],
        name="Toilets per Park",
        marker_color="blue",
    )
)

# Marker for toilets per acre
fig.add_trace(
    go.Scatter(
        x=park_toilet["BORO"],
        y=park_toilet["toilet_per_acre"],
        name="Toilets per Acre",
        mode="markers",
        marker=dict(size=12, color="red"),
        yaxis="y2",
    )
)

fig.update_layout(
    title="Comparison of Toilet Availability Metrics by Borough",
    xaxis_title="Borough",
    yaxis=dict(title="Toilets per Park"),
    yaxis2=dict(title="Toilets per Acre", overlaying="y", side="right"),
    template="plotly_white",
)

fig.show()
```

Finding 8: Manhattan stands out as the only borough where both indicators (toilets per park and toilets oer acre of park) are simultaneously the highest, indicating a well-distributed restroom infrastructure relative to both park numbers and parkland size. Meanwhile, Staten Island Sows the lowest values of both metrics, indicating that visitors may face significantly reduced access to public toilet despite the borough's large park areas
